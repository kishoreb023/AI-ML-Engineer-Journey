1.Imputation
2.Dropping

Importing library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns #data visualization
```

```python
#load dataser to pandas dataframe
dataset=pd.read_csv('/content/Placement_Dataset.csv')
```

```python
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

 imputation-replacing null values with statistical value like mean,median,mode



Central Tendencies


*  Mean-avg value/n
*   median-central avlue/avg of central value


*   mode-repeated value





```python
#analyse the distribution of data in salary
fig,ax=plt.subplots(figsize=(8,8))
sns.distplot(dataset.salary)
```

Skewed distribution in diagram  More outliers distribution more on one side

```python
#replace missing value with median value
```

```python
#converting into numeric data
dataset['salary'] = pd.to_numeric(dataset['salary'], errors='coerce')
```

```python
print(dataset['salary'].dtype)
```

```python
median_salary = dataset['salary'].median()

print(median_salary)
```

```python
dataset['salary'] = dataset['salary'].fillna(median_salary)
```

```python
dataset.isnull().sum()
```

```python
#filling null with mean value
mean_value=dataset['salary'].mean()
print(mean_value)

dataset['salary']=dataset['salary'].fillna(mean_value)
dataset.isnull().sum()
```

```python
#filling iwth mode value
mode_value=dataset['salary'].mode()
print(mode_value)

dataset['salary']=dataset['salary'].fillna(mode_value)
dataset.isnull().sum()
```
