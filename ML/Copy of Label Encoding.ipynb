Label Encoding->converting labels into numeric

```python
#importing

import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

Label Encoding of breast cancer dataset

```python
#loading data fro csv fiel to pandas dataframe

cancer_data=pd.read_csv('/content/breast_cancer_data.csv')
```

```python
cancer_data.head()
```

```python
cancer_data.shape
```

```python
#finding count of different labels
cancer_data['diagnosis'].value_counts()
```

```python
#load label encoder
label_encode=LabelEncoder()
```

Convert all the B AND M vaue to numeric value

```python
labels=label_encode.fit_transform(cancer_data.diagnosis)
```

```python
#appending labels to dataframe
cancer_data['target']=labels
cancer_data.head()
```

0--->Benin  1--->Malignant

```python
cancer_data['target'].value_counts()
```
